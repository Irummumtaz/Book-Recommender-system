```python
#Loading the book rrating datafame
ratings=pd.read_csv('BX-Book-Ratings.csv',sep=';',error_bad_lines=False,encoding='latin-1')
```

```python
ratings.head()
```

```python
#How many books user read 
ratings['user_id'].value_counts()
```

```python
#Users who had ratings more than 200 books
x=ratings['user_id'].value_counts()>200
x
```

```python
#Number of users rated more than 200
x[x].shape
```

```python
y=x[x].index
```

```python
y
```

```python
#Filter the user
ratings=ratings[ratings['user_id'].isin(y)]
```

```python
ratings.head()
```

```python
#Join ratings with books
rating_with_books=ratings.merge(books,on='ISBN')
```

```python
rating_with_books
```

```python
number_rating= rating_with_books.groupby('title')['rating'].count().reset_index()
```

```python
number_rating.head()
```

```python
number_rating.rename(columns={'rating':'num_of_rating'},inplace=True)
```

```python
number_rating.head()
```

```python
final_rating=rating_with_books.merge(number_rating,on='title')
```

```python
final_rating.head()
```

```python
#Books with atleast 50 ratings of user
final_rating=final_rating[final_rating['num_of_rating']>50]
```

```python
final_rating.head()
```

```python
#Drop Duplicates
final_rating.drop_duplicates(['user_id','title'],inplace=True)
```

```python
final_rating.isnull().sum()
```

```python
#Create pivot Table
book_pivot=final_rating.pivot_table(columns='user_id',index='title',values='rating')
```

```python
book_pivot
#NaN means 0 rating given here
```

```python
book_pivot.shape
```

```python
book_pivot.fillna(0,inplace=True)
```

```python
book_pivot
```

```python
#Convert pivot table to csr matrix
from scipy.sparse import csr_matrix
```

```python
book_sparse=csr_matrix(book_pivot)
```

```python
book_sparse
```

```python
type(book_sparse)
```

```python
from sklearn.neighbors import NearestNeighbors
model=NearestNeighbors(algorithm='brute')
```

```python
model.fit(book_sparse)
```

```python
book_pivot.iloc[237,:]
```

```python
distance,suggestion=model.kneighbors(book_pivot.iloc[237,:].values.reshape(1,-1),n_neighbors=6)
```

```python
distance
```

```python
suggestion
```
